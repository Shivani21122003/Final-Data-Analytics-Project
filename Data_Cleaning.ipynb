```python
 #Create a working copy
# Create a copy for cleaning
df_clean = df.copy()
# Check the original and working dataset sizes
print("Original dataset:", df.shape)
print("Working dataset:", df_clean.shape)
```

```python
#Handle missing values in age
# Calculate the median age
median_age = df_clean["age"].median()
print("Median age:", median_age)
```

```python
# Fill missing ages with the median
df_clean["age"] = df_clean["age"].fillna(median_age)
# Verify missing age values
print("Missing ages after cleaning:", df_clean["age"].isnull().sum())
```

```python
#Handle missing values in fare
# Calculate median fare
median_fare = df_clean["fare"].median()
print("Median fare:", median_fare)
```

```python
# Fill missing fare with median
df_clean["fare"] = df_clean["fare"].fillna(median_fare)
# Verify
print("Missing fares after cleaning:", df_clean["fare"].isnull().sum())
```

```python
#Handle missing values in embarked
# Find the most common embarkation port
most_common_port = df_clean["embarked"].mode()[0]
print("Most common port:", most_common_port)
```

```python
# Fill missing embarked values with the mode
df_clean["embarked"] = df_clean["embarked"].fillna(most_common_port)
# Verify
print("Missing embarked values after cleaning:",
      df_clean["embarked"].isnull().sum())
```

```python
#Review columns with many missing values
# Check missing values after our cleaning so far
remaining_missing = pd.DataFrame({
    "Missing Values": df_clean.isnull().sum(),
    "Missing Percentage": (
        df_clean.isnull().sum() / len(df_clean) * 100
    ).round(2)
})
remaining_missing[remaining_missing["Missing Values"] > 0]
```

```python
#Standardize text values
# Standardize text columns
text_columns = df_clean.select_dtypes(include=["object"]).columns

for col in text_columns:
    df_clean[col] = df_clean[col].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )
print("Text standardization completed.")
```

```python
#Validate the cleaned dataset
# 1. Check dataset dimensions
print("Dataset shape:", df_clean.shape)
# 2. Check duplicate rows
print("Duplicate rows:", df_clean.duplicated().sum())
# 3. Check missing values
print("\nMissing values in each column:")
print(df_clean.isnull().sum())
# 4. Check data types
print("\nData types:")
print(df_clean.dtypes)
```

```python
#Save your cleaned dataset
# Save cleaned dataset as CSV
df_clean.to_csv("titanic_cleaned.csv", index=False)
print("Cleaned CSV file saved successfully!")
```

```python
# Save cleaned dataset as Excel
df_clean.to_excel("titanic_cleaned.xlsx", index=False)
print("Cleaned Excel file saved successfully!")
```
